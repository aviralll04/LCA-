```python
#Assignment 4 
#addition  of matrix using arrays

matrix1 = [[1,2,3],[0,2,0],[1,2,0]]

matrix2 = [[3,2,1],[0,1,2],[0,4,5]]

result = [[0,0,0],[0,0,0],[0,0,0]]

#add corresponding elements
for i in range(len(matrix1)):
    for j in range(len(matrix1[0])):
        result[i][j] = matrix1[i][j] + matrix2[i][j]


        
print("matrix1")
for row in matrix1 :
    print(row)

print("matrix2")
for row in matrix2 :
    print(row)

    
print("Result")
for row in result : 
    print(row)
```
